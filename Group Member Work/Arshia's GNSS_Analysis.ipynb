# GNSS Timeseries Project

## Documentation overview
This notebook reads and displays velocity fields and time series from the NGF (EarthScope data center), UNR (University Nevada Reno), and JPL (Jet Propulsion Laboratory).  Most likely additional packages such as folium, pandas, numpy, ipywidgets, matplotlib, tabulate, geopy, json will need to be installed.  Conda installations from conda-forge should be possible for all packages.

The widget interface at the bottom of the notebook has three cells:

<b>Availability:</b> The “Latest NGF”, “Latest UNR”, and “Latest JPL” buttons must be used the first time the Notebook is run to create the data directory and subdirectories for each center that will used to save JSON files with information about all sites at the centers and time series files from each center as they are requested.  Latest updates should be selected to ensure the most up-to-date information is available.  The Availability check can used to see what is known about a 4-char code site name at each of the centers.

<b>Map:</b> The Map interface allows locations of sites, along with other sites within a user-specified radius to be plotted, along with 3-D velocities if available and selected.  Distance table for the 10 nearest sites to a site can be generated.

<b>Timeseries:</b> The Time series interface allows time series to be detrended, with optionally breaks from the EarthScope database removed.  Time series from different sites and centers can be overlayed.  Different backends can be used for plotting time series, with the ipympl backend allowing zooming and saving and the interactive one allowing identification of points.  The default inline backend should work on all systems.  The other backends may not work on all systems. 

All of the cells in the Notebook can be run, and the widgets and additional documentation will be at the bottom of the Notebook.  The code should run in a Jupyter Notebook or VSCode.

```python
def FitTS(xdata,ydata,sig):
    '''Fit times series.  Linear trend currently
    data is xdata time in days, ydata (NEU), sig
    https://en.wikipedia.org/wiki/Generalized_least_squares
    https://en.wikipedia.org/wiki/Reduced_chi-squared_statistic
    Outputs:
    array([grad, y-int]), [grad, grad_sd], [Wrms, chi, ndata]

    grad_sd is the formal sigma scaled by chi (NRMS) when chi > 1. It is still
    a white-noise formal error: GNSS position series carry flicker and
    random-walk noise that this two-parameter model does not represent, so
    grad_sd stays several times smaller than the velocity sigmas the data
    centres publish. Do not compare it with velsig, and do not use it as an
    error bar on a published velocity.
    '''

    # yr = xdata/365.25; # Years Since 2000/1/1
    # OmInv = 1./sig**2
    # ndata = int(xdata.size)  # number of data points
    # X = np.vstack((np.ones(ndata),yr)).T
    # Bvec = (X.T * OmInv) @ ydata.T
    # MCov = np.linalg.inv((X.T * OmInv) @ X)
    # MEst = MCov @ Bvec
    # Res = ydata - X @ MEst

    # chi = np.sqrt((Res.T @ (Res*OmInv))/(ndata-2))
    # wrms = np.sqrt(ndata/np.sum(OmInv))*chi

    # return np.flip(MEst), [MEst[1],np.sqrt(MCov[1,1])], [wrms,chi,ndata]

    # Set up to fit linear trend.  In later updates we could
    # more elaborate models with offsets, periodic and post-
    # seismic componensts.
    yr = xdata/365.25; # Years Since 2000/1/1
    wgh = 1./sig**2
    ndata = int(xdata.size)  # number of data points
    A = np.transpose([np.ones(ndata),yr])
    NormEQ = np.matmul(np.transpose(A)*wgh,A)
    Bvec = np.matmul(np.transpose(A)*wgh,np.transpose(ydata))
    MCov = np.linalg.inv(NormEQ)
    MEst = np.matmul(MCov,Bvec)
    Res = ydata-np.matmul(A,MEst)
    chi = np.sqrt(np.dot(np.transpose(Res),Res*wgh)/(ndata-2))
    wrms = np.sqrt(ndata/np.sum(wgh))*chi
    pfe = np.flip(MEst) # Set reverse order like polyfit
    #print("Est  : a=%5.3f, b=%5.3f" % tuple(MEst) )
    #print("Sig  : a=%5.3f, b=%5.3f" % tuple(np.sqrt(np.diag(MCov))) )
    #print('Fits',np.flip(MEst), [MEst[1],np.sqrt(MCov[1,1])], [wrms,chi,ndata])
    #print('xdata',xdata[0],xdata[-1],'ydata',ydata[0],ydata[-1])

    # The formal sigma from the normal matrix assumes white noise only. Scale
    # it by chi (NRMS) when the scatter exceeds the quoted sigmas, as
    # GLOBK-family software does. See the docstring for what this still misses.
    sigv = np.sqrt(MCov[1,1]) * max(float(chi), 1.0)

    return pfe, [MEst[1],sigv], [wrms,chi,ndata]

def detrended(xdata, ydata, sig):
    # m, c = FitTS(xdata, ydata, sig)[0]
    # return ydata - (m * xdata / 365.25 + c)
    pfe , Vel, Stat = FitTS(xdata, ydata, sig)
    # Return the residuals after removing linear trend and the 
    # Velocity estimate with sigma and the Statistics (WRMS, NRMS, #data)
    return ydata - (pfe[0] * xdata / 365.25 + pfe[1]), Vel, Stat


def remove_outliers_function(nd, ed, ud, ns, es, us, times, td, multiplier):
    bool_list = [all(abs(lst[i]-np.mean(lst)) <= multiplier*np.std(lst) for lst in (nd, ed, ud)) for i in range(len(nd))]
    nd_new = nd[bool_list]
    ed_new = ed[bool_list]
    ud_new = ud[bool_list]
    ns_new = ns[bool_list]
    es_new = es[bool_list]
    us_new = us[bool_list]
    times_new = times[bool_list]
    td_new = td[bool_list]
    return nd_new, ed_new, ud_new, ns_new, es_new, us_new, times_new, td_new

    
def calc_distance_earth(start_lat, start_lon, distance_km, direction = "east"):
    """Offsets a coordinate by a ground distance along one cardinal direction.

    Earth model: the WGS-84 ellipsoid, via geopy's geodesic. That is the one
    model used everywhere in this notebook that geometry is computed -- here,
    in draw_vector, in draw_ellipse, and in the d_km distances reported by the
    neighbour table -- and it matches gnss_core.velocity_endpoint in the main
    project and the WGS-84 coordinates the .vel files publish.

    One thing this does not control: ipl.Circle is rendered by Leaflet on its
    own 6371 km sphere, so the drawn radius circle can disagree with geodesic
    membership by up to ~0.25% (about 125 m on a 50 km radius). A station
    within roughly that distance of the boundary may render on the far side of
    the circle from where nearby_sites puts it.

    Args:
        start_lat: starting latitude in decimal degrees (WGS-84).
        start_lon: starting longitude in decimal degrees (WGS-84).
        distance_km: ground distance to travel, in kilometres.
        direction: one of "north", "south", "east", "west". Defaults to "east".

    Returns:
        A list [end_lat, end_lon] in decimal degrees.
    """
    bearing = {"north": 0.0, "east": 90.0, "south": 180.0, "west": 270.0}[direction]
    end = geodesic(kilometers=distance_km).destination((start_lat, start_lon), bearing)
    return [end.latitude, end.longitude]
    return [end_lat, end_lon]
```

```python
## GENERAL MAP OPERATIONS

d_km = lambda coords1, coords2: geopy.distance.geodesic(coords1, coords2).km
d_mi = lambda coords1, coords2: geopy.distance.geodesic(coords1, coords2).miles

def plot_locations(map, coordslist, tooltiplist, color="blue", dotrad=3,
                   on_click_callback=None, org=None):
    """Plots station markers on an ipyleaflet map with hover and click interactivity.

    Args:
        map: the ipyleaflet Map widget to add markers to.
        coordslist: list of (lat, lon) tuples for each marker position.
        tooltiplist: list of tooltip strings corresponding to each marker.
            Expected format is "SITEID, other info..." since the site ID
            is extracted by splitting on the first comma.
        color: fill and border color of the CircleMarker, defaults to "blue".
        dotrad: radius of the CircleMarker in pixels, defaults to 3.
        on_click_callback: legacy parameter kept for backwards compatibility,
            no longer used internally. Click behavior is handled by
            make_click_handler.
        org: organization string (e.g. "NGF", "UNR", "JPL") passed to hover
            and click handlers so they can look up site data in data_of[org].

    Returns:
        A list of ipyleaflet CircleMarker layers added to the map, for use
        by the layer registry so they can be removed individually later.
    """
    markers = []
    for coords, tooltip in zip(coordslist, tooltiplist):
        marker = ipl.CircleMarker(
            location=list(coords),
            title=tooltip,
            radius=dotrad,
            color=color,
            fill_color=color,
            fill_opacity=1,
            opacity=1,
        )

        # Extract site ID from tooltip string for panel lookup
        # e.g. "P096, vel: [-7.28, ...]" → "P096"
        siteid = tooltip.split(",")[0].strip()

        def make_click_handler(sid, o):
            """Returns a click handler captured at the current loop iteration.

            Args:
                sid: site ID string captured at the current loop iteration.
                o: organization string captured at the current loop iteration.

            Returns:
                A handler function that shows the site info panel when called.
            """
            def handler(**kwargs):
                show_site_panel(sid, o, click_info_output)
            return handler

        def make_hover_handler(sid, o):
            """Returns a hover handler captured at the current loop iteration.

            Updates the persistent info panel HTML in place when the mouse
            enters a marker. Only updates _site_info_html.value (a fast string
            assignment) rather than calling clear_output() and display(), which
            avoids flickering. The buttons in the panel remain untouched and
            always clickable regardless of hover activity.

            Uses _current_site as a dict with named keys "site_id" and "org"
            rather than a positional list, making the access self-documenting
            and less error-prone.

            Args:
                sid: site ID string captured at the current loop iteration.
                o: organization string captured at the current loop iteration.

            Returns:
                A handler function that updates the info panel when called.
            """
            def handler(**kwargs):
                # Update current site dict with named keys for clarity
                _current_site["site_id"] = sid
                _current_site["org"] = o
                if sid not in data_of.get(o, {}):
                    _site_info_html.value = f"<b>{sid}</b>"
                    return
                sd = data_of[o][sid]
                loc = sd.get("location", "N/A")
                ht_raw = sd.get("height", None)
                ht = [round(ht_raw, 3)] if ht_raw is not None else "N/A"
                vel_raw = sd.get("velocity", None)
                sig_raw = sd.get("velsig", None)
                vel = [round(v, 4) for v in vel_raw] if vel_raw else "N/A"
                sig = [round(s, 4) for s in sig_raw] if sig_raw else "N/A"
                _site_info_html.value = f"""
                    <b>{sid}</b> ({o})<br>
                    <b>Lat/Lon:</b> {loc[0]:.7f}, {loc[1]:.7f}<br>
                    <b>Height:</b> {ht}<br>
                    <b>Velocity NEU:</b> {vel}<br>
                    <b>Sigma NEU:</b> {sig}
                """
            return handler

        marker.on_click(make_click_handler(siteid, org))
        marker.on_mouseover(make_hover_handler(siteid, org))
        map.add_layer(marker)
        markers.append(marker)

    return markers


def nearby_sites(org, site, radius_in_km):
    """Finds all sites within a given radius of a reference site or coordinate.

    Args:
        org: organization key in data_of (e.g. "NGF", "UNR", "JPL").
        site: either a 4-character site ID string, a coordinate pair as a
            string like "(lat, lon)" or "[lat, lon]", or a list [lat, lon].
        radius_in_km: search radius in kilometers.

    Returns:
        A tuple of (id_coord_dist_list, cur_coords) where id_coord_dist_list
        is a list of (siteid, lat, lon, distance) tuples for all sites within
        the radius, and cur_coords is the resolved [lat, lon] of the reference
        site.
    """
    if isinstance(site, str) and site[0] in "([":
        site = list(eval(site))
    cur_coords = site
    if isinstance(site, str):
        cur_coords = data_of[org][site]["location"]
    assert not isinstance(cur_coords[0], str)
    id_coord_dist_list = []
    for siteid in data_of[org]:
        try:
            site_coords = data_of[org][siteid]["location"]
            distance = d_km(site_coords, cur_coords)
            if distance < radius_in_km:
                id_coord_dist_list.append((siteid, *site_coords, distance))
        except: pass
    return id_coord_dist_list, cur_coords


def basic_circle(map, center, radius_in_km):
    """Draws a reference radius circle on the map.

    Args:
        map: the ipyleaflet Map widget to add the circle to.
        center: [lat, lon] coordinates of the circle center.
        radius_in_km: radius of the circle in kilometers.
    """
    circle = ipl.Circle(
        location=list(center),
        radius=radius_in_km * 1000,  # ipyleaflet Circle radius is in metres
        color="black",
        weight=1,
        fill_opacity=0.0,
        opacity=1,
        fill_color="green",
        fill=True,
    )
    map.add_layer(circle)


def vec_add(c1, c2):
    """Computes element-wise addition of two coordinate vectors.

    Args:
        c1: first vector as a list of numbers.
        c2: second vector as a list of numbers.

    Returns:
        A list representing the element-wise sum of c1 and c2.
    """
    return [c1[i]+c2[i] for i in range(len(c1))]


def vec_sub(c1, c2):
    """Computes element-wise subtraction c1 - c2 of two coordinate vectors.

    Args:
        c1: first vector as a list of numbers.
        c2: second vector as a list of numbers.

    Returns:
        A list representing the element-wise difference c1 - c2.
    """
    return [c1[i]-c2[i] for i in range(len(c1))]


def draw_vector(map, center, direction, scale, color="black", label="None"):
    """Draws a velocity vector arrow on the map with a precise geographic arrowhead.

    The entire arrow (shaft and both arrowhead arms) is drawn as a single
    Polyline to minimize add_layer calls and improve performance with large
    numbers of stations. Arrowhead arms are computed using calc_distance_earth
    so they are equal length in real-world distance regardless of latitude,
    avoiding the cos(lat) distortion that occurs in degree-space. Arrowhead
    size is 8% of vector length.

    Args:
        map: the ipyleaflet Map widget to draw the vector on.
        center: [lat, lon] of the vector origin (station location).
        direction: [north, east, up] velocity components in mm/yr.
        scale: scaling factor converting mm/yr to km on the map.
        color: color of the vector line, defaults to "black". Passed as
            "rgb(r,0,b)" from plot_velocity for vertical velocity color coding.
        label: text label to display at the vector tip. Pass "None" to suppress,
            defaults to "None".

    Returns:
        A tuple of (endpoint, layers) where endpoint is the [lat, lon] of the
        vector tip (used by plot_velocity to center the error ellipse), and
        layers is a list of ipyleaflet layers added to the map for use by the
        layer registry.
    """
    # One geodesic step along the true bearing of the velocity, matching
    # gnss_core.velocity_endpoint in the main project. Offsetting latitude and
    # longitude independently (the previous approach) is a small-angle
    # approximation that also disagreed with its own measured length.
    vec_len_km = scale * math.hypot(direction[0], direction[1])
    bearing = math.degrees(math.atan2(direction[1], direction[0]))
    tip = geodesic(kilometers=vec_len_km).destination(tuple(center), bearing)
    endpoint = [tip.latitude, tip.longitude]
    layers = []

    head_len_km = vec_len_km * 0.08

    if head_len_km < 0.01:
        line = ipl.Polyline(locations=[list(center), endpoint],
                            weight=2, color=color, fill=False)
        map.add_layer(line)
        layers.append(line)
        if label != "None":
            label_marker = ipl.Marker(
                location=endpoint,
                icon=ipl.DivIcon(
                    html='<div style="font-size:12pt;white-space:nowrap;">'+label+'</div>',
                    icon_size=[250, 36],
                    icon_anchor=[0, 0],
                ),
                interactive=False
            )
            map.add_layer(label_marker)
            layers.append(label_marker)
        return endpoint, layers


    # Compute both arrowhead arm endpoints at ±150° from vector bearing
    # Using tuples in the loop since side_deg values are fixed constants
    # that should not be modified
    arm_points = []
    for side_deg in (150, -150):  # tuple — fixed constants, not mutable
        arm_bearing = (bearing + side_deg) % 360
        arm_point = geodesic(kilometers=head_len_km).destination(
            tuple(endpoint), arm_bearing)
        arm_lat, arm_lon = arm_point.latitude, arm_point.longitude
        arm_points.append((arm_lat, arm_lon))  # tuple — fixed coordinate pair

    # Draw entire arrow as one Polyline: start → tip → arm1 → tip → arm2
    # This uses 1 add_layer call instead of 3, improving performance
    arrow = ipl.Polyline(
        locations=[
            list(center),
            endpoint,
            list(arm_points[0]),  # convert tuple back to list for ipyleaflet
            endpoint,
            list(arm_points[1])
        ],
        weight=2, color=color, fill=False
    )
    map.add_layer(arrow)
    layers.append(arrow)

    if label != "None":
        label_marker = ipl.Marker(
            location=endpoint,
            icon=ipl.DivIcon(
                html='<div style="font-size:12pt;white-space:nowrap;">'+label+'</div>',
                icon_size=[250, 36],
                icon_anchor=[0, 0],
            ),
            interactive=False
        )
        map.add_layer(label_marker)
        layers.append(label_marker)

    return endpoint, layers


def draw_ellipse(map, center, sigma_n, sigma_e, confidence, vel_scale, color="black",
                 correlation=0.0):
    """Draws a horizontal-velocity confidence ellipse at the vector tip.

       The ellipse is the k-sigma contour of the 2x2 North/East covariance
       C = [[sigma_n**2, rho*sigma_n*sigma_e], [rho*sigma_n*sigma_e, sigma_e**2]]
       with k = sqrt(-2 * log(1 - confidence)). The semi-axes are
       k * sqrt(eigenvalue) * vel_scale and the major axis is rotated to the
       eigenvector, so the ellipse tilts whenever a correlation is supplied.
       With rho = 0 the eigenvalues are sigma_n**2 and sigma_e**2 and the axes
       lie along North and East, so behaviour is unchanged without correlation.

       The confidence levels are 2-D coverage, not the 1-D 68/95/99.7 values:
       k = 1, 2, 3 enclose 39.35%, 86.47% and 98.89% of a bivariate normal.

    Args:
        map: the ipyleaflet Map widget to draw the ellipse on.
        center: [lat, lon] coordinates of the ellipse center (vector tip).
        sigma_n: 1-sigma uncertainty in the north component in mm/yr.
        sigma_e: 1-sigma uncertainty in the east component in mm/yr.
        confidence : Confidence level as a fraction (0.3935, 0.8647, 0.95,
                     0.98889); passed from ellipse_confidence.value.
        vel_scale: scale factor converting mm/yr to km on the map. Pass the
               SAME km per mm/yr the arrow uses, i.e. velocity_scale_input
               times an ellipse_scale_input of 1. The ellipse is only readable
               as the uncertainty of the arrow while the two share a scale;
               any other multiplier is presentation only.
        color: outline color of the ellipse, defaults to "black".
        correlation: North/East correlation coefficient rho, defaults to 0.0.
               Only NGF publishes it (field Rne); UNR and JPL records have no
               correlation, so their ellipses stay axis-aligned.

    Returns:
        The ipyleaflet Polyline layer added to the map, or None if the ellipse
        is too small to display. Used by the layer registry for individual
        layer removal.
    """
    # psvelo formula: sqrt(-2 * log(1 - confidence)). This is 2-D coverage, so
    # k = 1, 2, 3 correspond to 39.35%, 86.47% and 98.89%, not 68/95/99.7.
    confidence_scale = math.sqrt(-2 * math.log(1 - confidence))

    # Closed-form eigen-decomposition of the symmetric 2x2 covariance ordered
    # (North, East):  C = [[var_n, cov_ne], [cov_ne, var_e]]
    var_n = sigma_n ** 2
    var_e = sigma_e ** 2
    cov_ne = correlation * sigma_n * sigma_e
    mean_var = 0.5 * (var_n + var_e)
    root = math.hypot(0.5 * (var_n - var_e), cov_ne)
    # max(..., 0.0) guards round-off and any |correlation| > 1 in the source.
    semi_major_km = confidence_scale * vel_scale * math.sqrt(max(mean_var + root, 0.0))
    semi_minor_km = confidence_scale * vel_scale * math.sqrt(max(mean_var - root, 0.0))
    # Azimuth of the major axis from North towards East. Positive correlation
    # tilts the ellipse into the NE-SW quadrant, negative into NW-SE.
    azimuth = 0.5 * math.atan2(2 * cov_ne, var_n - var_e)

    # Nothing is visible unless the LARGEST axis clears the threshold. The old
    # 'radius_n and radius_e' test was only meaningful while the axes were
    # locked to north and east.
    if semi_major_km < 0.001:
        return None

    # Vectorized numpy computation of all 37 ellipse points at once
    # calc_distance_earth calls with 2 numpy array operations
    angles = np.radians(np.arange(0, 361, 10))

    # Rotate the parametric ellipse out of its own axes into (north, east) km.
    # At angles=0 this is the semi-major point, which lies along the azimuth.
    cos_az = math.cos(azimuth)
    sin_az = math.sin(azimuth)
    north_km = (semi_major_km * np.cos(angles) * cos_az
                - semi_minor_km * np.sin(angles) * sin_az)
    east_km = (semi_major_km * np.cos(angles) * sin_az
               + semi_minor_km * np.sin(angles) * cos_az)

    # Place every point with the same WGS-84 geodesic step draw_vector uses,
    # so the ellipse and the arrow it decorates sit on one Earth model.
    points = []
    for n_km, e_km in zip(north_km, east_km):
        point = geodesic(kilometers=math.hypot(n_km, e_km)).destination(
            tuple(center), math.degrees(math.atan2(e_km, n_km)))
        points.append([point.latitude, point.longitude])

    ellipse = ipl.Polyline(
        locations=points,
        color=color,
        weight=1,
        fill=False,
    )
    map.add_layer(ellipse)
    return ellipse
```

```python
# Functions for editing and plotting time series.
def remove_brac(string):
    '''
    "P123 (UNR)" --> ("P123", "UNR")
    '''
    i=-1
    while string[i] != '(': i-=1
    return string[:i-1], string[i+1:-1]


def plot_ts_graph(siteid, org, Update, detrend, ax0, ax1, ax2, yearrange, breaks, errorbars, errorbar_outline, outlier, shift, color):
    '''
    Adding a single plot to the original plot defined in plot_ts_graph_list()

    When outliers are removed, the velocity sigma reported in the returned
    data frame comes from the final fit, so it refers to the same data as the
    WRMS, chi and Num printed beside it.

    The sigma of a removed break offset is NOT propagated. An offset error is
    one draw shared by the whole post-break segment, not independent noise per
    epoch, and FitTS weights each epoch independently; adding it in quadrature
    to every post-break sigma would understate its effect on the velocity
    while appearing to account for it, and would also re-weight the fit and
    move the velocity itself. Representing it properly needs a common nuisance
    parameter, which this two-parameter model does not have.
    '''

    try:
        times, tseries = Read[org](siteid, Update)
    except:
        print('Error getting',siteid,'from',org)
        times = []; tseries = []
        return
 
    td, nd, ns, ed, es, ud, us = tseries

    ## restricting timeseries to yearrange
    # find the start and end index
    start, end = 0, len(td)
    while start < end and times[start] < yearrange[0]: start+=1
    while end > start and times[end-1] > yearrange[1]: end-=1
    if start >= end:
        print('No data for',siteid,'from',org,'in the selected date range')
        return
    # restrict

    # for name in [times, td, nd, ns, ed, es, ud, us]:
    #     name = name[start:end]
    times, td = times[start:end], td[start:end]
    nd, ns = nd[start:end], ns[start:end]
    ed, es = ed[start:end], es[start:end]
    ud, us = ud[start:end], us[start:end]
    
    if breaks:
        if "breaks" in data_of[org][siteid]:
            breaktimes_dict = {}
            for dt in data_of[org][siteid]["breaks"]:
                try:
                    yr_mm_day = [int(date) for date in dt[1:-1].split(",")][0:5]
                    breaktime = datetime(*eval(str(yr_mm_day)))
                    breaktimes_dict[breaktime] = [data_of[org][siteid]["breaks"][dt]["offsets"], int(np.where(times >= breaktime)[0][0])]
                except:
                    pass
            # [45.01, 0.67, 39.72, 0.59, -27.67, 1.83], [dN (mm), sN (mm), dE (mm), sE (mm), dU (mm), sU (mm)]
            interval_counter = 0
            breaktimes_list = list(breaktimes_dict)
            ndat = len(ud)
            while interval_counter in range(len(breaktimes_dict)):
                dict_key = breaktimes_list[interval_counter]
                interval_start = breaktimes_dict[dict_key][1]
                # Using -1 as last point skipped last point
                nd[interval_start:ndat] = nd[interval_start:ndat] - breaktimes_dict[dict_key][0][0]
                ed[interval_start:ndat] = ed[interval_start:ndat] - breaktimes_dict[dict_key][0][2]
                ud[interval_start:ndat] = ud[interval_start:ndat] - breaktimes_dict[dict_key][0][4]
                interval_counter +=1
        else:
            print("No breaks information!")

    dfa = []
    if detrend:
        nd, VelN, StatN = detrended(td, nd, ns)
        ed, VelE, StatE = detrended(td, ed, es)
        ud, VelU, StatU = detrended(td, ud, us)

        if outlier: # while length of the array before and after are different, keep iterating the code
            initial_length = len(nd)
            length1 = 1
            length2 = 2
            iteration_cnt = 0
            while length1 != length2:
                # print(f"length of nd: {len(nd)}, {len(ud)}, {(len(ed))}, {len(td)}")
                # print(f"length of s: {len(ns)}, {len(us)}, {(len(es))}")
                length1 = len(nd)
                nd, ed, ud, ns, es, us, times, td = remove_outliers_function(nd, ed, ud, ns, es, us, times, td, outlier)
                # print(f"lengths after: {len(nd)}, {len(ud)}, {(len(ed))}, {len(td)}")
                # print(f"before: {np.max(ud)}, {np.min(ud)}")
                nd, dVelN, StatN = detrended(td, nd, ns)
                VelN[0] = VelN[0]+dVelN[0]   # Fit is to residuals, so update total
                # The iterated velocity equals a single fit on the retained
                # data, so the sigma, WRMS, chi and Num must all come from
                # that same final fit rather than from the first pass.
                VelN[1] = dVelN[1]           # sigma belongs to the final fit

                ed, dVelE, StatE = detrended(td, ed, es)
                VelE[0] = VelE[0]+dVelE[0]   # Fit is to residuals, so update total
                VelE[1] = dVelE[1]           # sigma belongs to the final fit
                ud, dVelU, StatU = detrended(td, ud, us)
                VelU[0] = VelU[0]+dVelU[0]   # Fit is to residuals, so update total
                VelU[1] = dVelU[1]           # sigma belongs to the final fit

                length2 = len(nd)
                iteration_cnt +=1
                # print(f"after: {np.max(ud)}, {np.min(ud)}")
            print(f"Final {len(nd)} data; Number of outliers removed: {initial_length - len(nd)} with {iteration_cnt} iterations: ")

        # Now save the data frame information for the fit.
        VelNEU = VelN+VelE+VelU ; StatNEU = StatN[0:2]+StatE[0:2]+StatU
        labl = str(siteid)+'-'+str(org)
        # print('DFrame ',VelNEU,StatNEU,labl)
        dfa = pd.DataFrame([VelNEU+StatNEU], \
            columns=['Vn','σVn','Ve','σVe','Vu','σVu','WRMS N','χn','WRMS E','χe','WRMS U','χu','Num'],index=[labl])


    nd = nd+shift ;  ed = ed+shift ; ud = ud+shift*3

    if errorbars[0]:
        ax0.errorbar(times,nd,yerr=[ns,ns],errorevery=errorbars[1], capsize=2, color = color, linewidth=1) #, ecolor='black')
        ax1.errorbar(times,ed,yerr=[es,es],errorevery=errorbars[1], capsize=2, color = color, linewidth=1) #, ecolor='black')
        ax2.errorbar(times,ud,yerr=[us,us],errorevery=errorbars[1], capsize=2, color = color, linewidth=1) #, ecolor='black')

    if errorbar_outline[0]:
        for ax, d, s in [(ax0, nd, ns), (ax1, ed, es), (ax2, ud, us)]:
            ax.fill(list(times) + list(reversed(times)),
                list(d+s) + list(np.flip(d-s)),
                alpha=errorbar_outline[1], linewidth=1, color = color, label="_"+siteid)

    for ax, d in [(ax0, nd), (ax1, ed), (ax2, ud)]:
        ax.plot(times, d, linewidth=0.5, label=siteid + " (" + org + ")", color = color)
        if "breaks" in data_of[org][siteid]:
            for dt in data_of[org][siteid]["breaks"]:
                yr_mm_day = [int(date) for date in dt[1:-1].split(",")][0:3]
                breaktime = datetime(*eval(str(yr_mm_day)))
                try:
                    np.where(times >= breaktime)[0][0]
                    ax.axvline(x=breaktime, color='r', ls='--')
                except:
                    pass

    return dfa

def plot_ts_graph_list(idlist, Update, yearrange, breaks, detrend, errorbars, errorbar_outline, outlier, resolution = "Low Res", shift=0):
    '''
    Plots the timeseries of a list of sites.
    '''
    global tsfig
    tsfig = plt.figure(figsize=(15, 10), dpi = {"HD": 600, "Regular": 300, "Low Res": 100}[resolution])
    gs = gridspec.GridSpec(3, 1, height_ratios=[1, 1, 1])

    ax0 = plt.subplot(gs[0])
    ax0.set_ylabel("ΔNorth (mm)")
    ax0.yaxis.set_tick_params(labelrotation=90)
    plt.setp(ax0.get_xticklabels(), visible=False)

    ax1 = plt.subplot(gs[1], sharex = ax0)
    ax1.set_ylabel("ΔEast (mm)")
    ax1.yaxis.set_tick_params(labelrotation=90)
    plt.setp(ax1.get_xticklabels(), visible=False)

    ax2 = plt.subplot(gs[2], sharex = ax0)
    ax2.set_ylabel("ΔUp (mm)")
    ax2.yaxis.set_tick_params(labelrotation=90)
    ax2.set_xlabel("Time")

    colors = ["b", 'g', 'r', 'c', 'm', 'y', 'k', 'sienna']
    color_count = 0
    dfall = None
    for idorg in idlist:
        dfl = plot_ts_graph(*remove_brac(idorg), Update, detrend, ax0, ax1, ax2, yearrange, breaks, errorbars, errorbar_outline, outlier, shift[idorg] if idorg in shift else 0, colors[color_count])
        color_count = (color_count + 1) % len(colors)

        # Create a pandas data frame will the results in it
        if detrend and dfl is not None :
            if dfall is None :
                dfall = dfl
            else:
                dfall = pd.concat([dfall,dfl])
 
    # When all done display the data frame
    if detrend and dfall is not None :
        # Format the DataFrame (Early versions of pandas may have problems)
        display(dfall.style.format({'Vn': '{:.2f}','σVn': '{:.3f}',
                                  'Ve': '{:.2f}','σVe': '{:.3f}',
                                  'Vu': '{:.2f}','σVu': '{:.3f}',
                                  'WRMS N': '{:.2f}','χn': '{:.2f}',
                                  'WRMS E': '{:.2f}','χe': '{:.2f}',
                                  'WRMS U': '{:.2f}','χu': '{:.2f}','Num':'{:d}'}))
 
        #display(dfall.style.format('{:.2f}')) 
        MD = False # Set true to get MarkDown formatted table.
        if MD :
            headers = ['Site','Vn','σVn','Ve','σVe','Vu','σVu','WRMS N','χn','WRMS E','χe','WRMS U','χu','Num']
            table = dfall.to_numpy()
            index = np.array(list(dfall.index.values))
            fulltable = np.c_[index , table]
            print('\nPIPE: For pandas.to_markdown')
            print(tabulate(fulltable, headers, tablefmt="pipe",floatfmt=("%s",".2f", ".3f",".2f",".3f",".2f", ".3f",".2f",".2f",".2f", ".2f",".2f",".2f",".2f")))


    # leg = ax0.legend(bbox_to_anchor=(1, 1))
    leg = ax0.legend()

    plt.subplots_adjust(hspace=.0)

    #for legobj in leg.legend_handles: # The lines in the legend were too thin previously.
    #    legobj.set_linewidth(3.0)

    plt.show()
    #return tsfig
```

```python
# Map Functions
# Choices of maps come from https://leaflet-extras.github.io/leaflet-providers/preview/

# Mutable list storing the current TileLayer so the basemap can be swapped
# without rebuilding the map. A list is used so the reference can be updated
# in place across function scopes (same pattern as map[0]).
current_tile_layer = [None]

# Registry mapping each "SITEID, ORG, RADIUSkm" entry to its list of map layers.
# Allows individual site layers to be removed from the map without closing it.
layer_registry = {}

# Persistent site info panel — all widgets defined once at block load time and
# never recreated. Only _site_info_html.value is updated on hover or click,
# which is a fast string assignment that avoids flickering and keeps the
# buttons always wired and clickable.

# _current_site is a dict with named keys rather than a positional list.
# Named keys make access self-documenting and less error-prone than
# relying on index positions like [0] and [1].
_current_site = {"site_id": None, "org": None}

_site_info_html = wg.HTML(value="")
_add_button = wg.Button(description="Add to Timeseries", layout=wg.Layout(width='180px'))
_plot_button = wg.Button(description="Plot Now", layout=wg.Layout(width='180px'),
                         button_style='success')


def _do_add(b):
    """Adds the currently selected site to the time series list.

    Reads _current_site dict to determine which site to add, so it always
    acts on whichever site was most recently hovered over or clicked on the
    map.

    Args:
        b: button click event, not used directly.
    """
    if _current_site["site_id"] is None:
        return
    # Read from named dict keys for clarity
    s = _current_site["site_id"]
    o = _current_site["org"]
    extended = s + " (" + o + ")"
    if extended not in ts_sites.options:
        ts_sites.options = list(ts_sites.options) + [extended]
    with timeseries_output:
        display(wg.HTML(f'<em style="color:green">Added {extended} to list</em>'))


def _do_plot(b):
    """Adds the currently selected site to ts_sites and plots it immediately.

    Adds the current site to ts_sites if not already present, selects it so
    list_to_graph sees it as the active selection, then calls list_to_graph
    to plot immediately.

    Args:
        b: button click event, not used directly.
    """
    if _current_site["site_id"] is None:
        return
    # Read from named dict keys for clarity
    s = _current_site["site_id"]
    o = _current_site["org"]
    extended = s + " (" + o + ")"
    if extended not in ts_sites.options:
        ts_sites.options = list(ts_sites.options) + [extended]
    ts_sites.value = (extended,)
    list_to_graph(None)


# Wire buttons once — they stay wired for the lifetime of the notebook
_add_button.on_click(_do_add)
_plot_button.on_click(_do_plot)

# Display the persistent panel structure once into click_info_output.
# Subsequent hover and click events only update _site_info_html.value,
# never recreating or clearing these widgets.
with click_info_output:
    display(wg.VBox([_site_info_html, wg.HBox([_add_button, _plot_button])]))


def show_site_panel(site_id, org, panel):
    """Updates the persistent info panel with data for the selected site.

    Updates _current_site dict and _site_info_html.value in place so the
    buttons always act on the correct site without any widget recreation or
    output clearing.

    Args:
        site_id: 4-character GNSS station identifier.
        org: organization key in data_of (e.g. "NGF", "UNR", "JPL").
        panel: the ipywidgets Output widget to display into (click_info_output).
    """
    # Update named dict keys so _do_add and _do_plot always act on correct site
    _current_site["site_id"] = site_id
    _current_site["org"] = org

    if site_id not in data_of.get(org, {}):
        _site_info_html.value = f"<b>{site_id}</b>"
        return

    site_data = data_of[org][site_id]
    loc = site_data.get("location", "N/A")
    ht_raw = site_data.get("height", None)
    ht = [round(ht_raw, 3)] if ht_raw is not None else "N/A"
    vel_raw = site_data.get("velocity", None)
    sig_raw = site_data.get("velsig", None)
    vel = [round(v, 4) for v in vel_raw] if vel_raw else "N/A"
    sig = [round(s, 4) for s in sig_raw] if sig_raw else "N/A"

    # Update HTML string in place — no widget recreation needed
    _site_info_html.value = f"""
        <b>{site_id}</b> ({org})<br>
        <b>Lat/Lon:</b> {loc[0]:.7f}, {loc[1]:.7f}<br>
        <b>Height:</b> {ht}<br>
        <b>Velocity NEU:</b> {vel}<br>
        <b>Sigma NEU:</b> {sig}
    """


def get_basemap_layer(b):
    """Builds and returns an ipyleaflet TileLayer for the selected basemap.

    In Folium, the basemap was passed as constructor kwargs to Map() at
    creation time, requiring a full map rebuild to change it. In ipyleaflet,
    basemaps are TileLayer objects that can be added and removed from the live
    map without touching any other layers.

    Args:
        b: button click event, not used directly.

    Returns:
        An ipyleaflet TileLayer ready to add to the map via map.add_layer().
    """
    attr = (
        '&copy; <a href="https://www.openstreetmap.org/copyright">OpenStreetMap</a> '
        'contributors, &copy; <a href="https://cartodb.com/attributions">CartoDB</a>'
        'contributors, <a href="http://viewfinderpanoramas.org">SRTM</a>'
    )
    if BaseMap_butt.value == 'OpenTopo':
        tiles = "https://{s}.tile.opentopomap.org/{z}/{x}/{y}.png"
    elif BaseMap_butt.value == 'OpenStreetMap':
        tiles = "https://tile.openstreetmap.org/{z}/{x}/{y}.png"
    else:
        tiles = "https://server.arcgisonline.com/ArcGIS/rest/services/World_Imagery/MapServer/tile/{z}/{y}/{x}"
        attr = 'Tiles &copy; Esri &mdash; Source: Esri, i-cubed, USDA, USGS, AEX, GeoEye, Getmapping, Aerogrid, IGN, IGP, UPR-EGP, and the GIS User Community'
    return ipl.TileLayer(url=tiles, attribution=attr)


def swap_basemap(b):
    """Swaps the active basemap tile layer without rebuilding the map.

    Removes the currently stored TileLayer and adds the newly selected one,
    preserving all station markers, circles, vectors, and ellipses already
    on the map.

    Args:
        b: button click event, not used directly.
    """
    if isinstance(map[0], ipl.Map):
        if current_tile_layer[0] is not None:
            map[0].remove_layer(current_tile_layer[0])
        new_layer = get_basemap_layer(b)
        current_tile_layer[0] = new_layer
        map[0].add_layer(new_layer)

BaseMap_Set.on_click(swap_basemap)


def new_map(b):
    """Creates a new ipyleaflet Map and displays it in map_output.

    Centers on the first selected site in map_radius_list if one exists,
    otherwise defaults to (40, -100) at zoom 4 (continental US view).
    Removes the default OSM tile layer ipyleaflet adds automatically and
    replaces it with the user-selected basemap.

    Args:
        b: button click event, not used directly.
    """
    map_output.clear_output()

    ll = [40, -100]; zoom = 4
    for option in list(map_radius_list.value):
        option = option.split(", ")
        ll = data_of[option[1]][option[0]]["location"]
        zoom = 6

    map[0] = ipl.Map(
        center=list(ll),
        zoom=zoom,
        scroll_wheel_zoom=True,
        layout=wg.Layout(width='1000px', height='600px')
    )

    # Remove default OSM TileLayer ipyleaflet adds automatically at Map creation
    for layer in list(map[0].layers):
        if isinstance(layer, ipl.TileLayer):
            map[0].remove_layer(layer)

    # Add user-selected basemap and store reference for later swapping
    tile_layer = get_basemap_layer(b)
    current_tile_layer[0] = tile_layer
    map[0].add_layer(tile_layer)

    # Add scale bar — in Folium this was control_scale=True on Map()
    # In ipyleaflet controls are added separately with add_control()
    map[0].add_control(ipl.ScaleControl(position='bottomleft'))

    # Display map once — after this the live widget updates automatically
    with map_output:
        display(map[0])

new_map_butt.on_click(new_map)


def reload_map(b):
    """Forces a full redisplay of the map widget.

    In ipyleaflet the map updates in place so this is rarely needed. Kept
    as a fallback in case the output cell gets cleared or scrolled away.

    Args:
        b: button click event, not used directly.
    """
    if isinstance(map[0], ipl.Map):
        map_output.clear_output()
        with map_output:
            display(map[0])

reload_map_butt.on_click(reload_map)


def close_map(b):
    """Clears all map-related outputs and resets the map and panel state.

    Resets current_tile_layer, _current_site, and layer_registry so stale
    references do not carry over to the next map session.

    Args:
        b: button click event, not used directly.
    """
    map_output.clear_output()
    nearest_site_output.clear_output()
    graph_output.clear_output()
    _site_info_html.value = ""
    # Reset named dict keys to None rather than reassigning the whole dict,
    # so all references to _current_site across the notebook stay valid
    _current_site["site_id"] = None
    _current_site["org"] = None
    layer_registry.clear()
    map[0] = "Not initialized yet!"
    current_tile_layer[0] = None

close_map_butt.on_click(close_map)


def map_click_to_ts(site_id, org):
    """Adds a site to the time series list programmatically.

    Formats the site as "SITEID (ORG)" to match the format used throughout
    the time series panel in blocks 8, 14, and 16.

    Args:
        site_id: 4-character GNSS station identifier.
        org: organization key (e.g. "NGF", "UNR", "JPL").
    """
    extended = site_id + " (" + org + ")"
    if extended not in ts_sites.options:
        ts_sites.options = list(ts_sites.options) + [extended]
    with timeseries_output:
        display(wg.HTML(f'<em style="color:green">Added {extended} to time series list</em>'))


####################
# SITE/RADIUS LIST #
####################

def update_map_list(b):
    """Adds a site/org/radius entry to map_radius_list if the site ID is valid.

    Entries are stored as comma-separated strings in the format
    "SITEID, ORG, RADIUSkm", which are parsed back out by site_circle
    and nn_graph.

    Args:
        b: button click event, not used directly.
    """
    siteid = site_id.value
    radius = site_radius.value
    org = org_map_select.value
    if siteid in data_of[org].keys():
        addition = f"{siteid}, {org}, {radius}km"
        # Duplicate check prevents the same entry appearing twice if the
        # block is re-run or the button is accidentally clicked twice
        if addition not in map_radius_list.options:
            map_radius_list.options = list(map_radius_list.options) + [addition]
    else:
        print("Not A Valid Site!")

add_sites_map_button.on_click(update_map_list)


def remove_map_list(b):
    """Removes selected entries from map_radius_list and their map layers.

    Uses the layer registry to remove all ipyleaflet layers associated with
    each selected entry so the map updates immediately without closing it.

    Args:
        b: button click event, not used directly.
    """
    options = list(map_radius_list.options)
    for option in map_radius_list.value:
        options.remove(option)
        if option in layer_registry:
            for layer in layer_registry[option]:
                try:
                    map[0].remove_layer(layer)
                except Exception:
                    pass
            del layer_registry[option]
    map_radius_list.options = options

remove_sites_map_button.on_click(remove_map_list)


################
# SITE BUTTONS #
################

def site_circle(b):
    """Plots all sites and radii in map_radius_list onto the map.

    Clears all previously tracked layers before redrawing so the map always
    matches the current contents of map_radius_list exactly. Tracks all layers
    added per site entry in layer_registry so they can be removed individually
    without closing the map. Optionally plots velocity vectors and error
    ellipses for each site depending on the state of plot_vec_check and
    plot_ellipse_check.

    Args:
        b: button click event, or None when called programmatically.
    """
    if not isinstance(map[0], ipl.Map):
        new_map(b)

    # Remove all existing tracked layers before redrawing so the map always
    # matches the current contents of map_radius_list
    for entry_key in list(layer_registry.keys()):
        for layer in layer_registry[entry_key]:
            try:
                map[0].remove_layer(layer)
            except Exception:
                pass
    layer_registry.clear()

    # Parse map_radius_list entries into a dict keyed by site ID
    # Dict is appropriate here since we look up entries by site ID
    selected_options = {}
    for option in list(map_radius_list.options):
        option = option.split(", ")
        selected_options[option[0]] = {"org": option[1], "radius": option[2]}

    plotvec = plot_vec_check.value

    for siteid in selected_options:
        org = selected_options[siteid]["org"]
        radius = int(selected_options[siteid]["radius"][0:-2])
        entry_key = f"{siteid}, {org}, {radius}km"
        entry_layers = []

        # Plot the reference site marker in black
        # org=org is passed so hover and click handlers can look up site data
        ref_markers = plot_locations(
            map[0],
            [data_of[org][siteid]["location"] if isinstance(siteid, str) else siteid],
            [f"{siteid}, no vel. data"] if not isinstance(get_vel(siteid, org), tuple)
            else [f"{siteid}, vel:{get_vel(siteid, org)[0]}"],
            color="black", dotrad=4, org=org)
        entry_layers.extend(ref_markers)

        if plotvec:
            vec_layers = plot_velocity(
                siteid, org,
                list(eval(vel_siglim_form.value)) if vel_siglim_form.value else [1, 1, 2])
            entry_layers.extend(vec_layers)

        if radius > 0:
            site_list, center = nearby_sites(org, siteid, radius)
            circle = ipl.Circle(
                location=list(center),
                radius=radius * 1000,
                color="black", weight=1,
                fill_opacity=0.0, opacity=1,
                fill_color="green", fill=True)
            map[0].add_layer(circle)
            entry_layers.append(circle)

            # Build tooltip strings for nearby sites including velocity if available
            # site_name_list uses a tuple comprehension since it is read-only
            site_name_list = tuple(el[0] for el in site_list)
            tooltip_list = []
            for site in site_name_list:
                if isinstance(get_vel(site, org), tuple):
                    velstr = [f"{x:.2f}" for x in get_vel(site, org)[0]]
                    velflt = [float(x) for x in velstr]
                    ll = data_of[org][site]['location']
                    ht = data_of[org][site]['height']
                    locstr = f"{ll[0]:.5f} {ll[1]:.5f} {ht:.3f}"
                    allstr = f"{site}, vel: {velflt}, loc: {locstr}"
                    tooltip_list.append(allstr)
                else:
                    tooltip_list.append(f"{site}, no vel. data")

            # Plot nearby site markers
            nearby_markers = plot_locations(
                map[0], [el[1:3] for el in site_list], tooltip_list, org=org)
            entry_layers.extend(nearby_markers)

            if plotvec:
                # Plot velocity vectors, decimated by thin_velocities_input
                counter = 0
                for el in site_list:
                    if counter == 0:
                        vec_layers = plot_velocity(
                            el[0], org,
                            list(eval(vel_siglim_form.value)) if vel_siglim_form.value else [1, 1, 2])
                        entry_layers.extend(vec_layers)
                    counter += 1
                    if counter == thin_velocities_input.value:
                        counter = 0

                # Scale arrow commented out — partner's widget handles this
                # vsloc = np.copy(data_of[org][siteid]["location"])
                # vsrad = site_radius.value
                # vsloc[0] = vsloc[0] + 1.1 * vsrad / 111.0
                # vsvel = [0, arrow_length_input.value, 0]
                # vslab = f"{vsvel[1]:.0f} mm/yr"
                # _, scale_layers = draw_vector(
                #    map[0], vsloc, vsvel, velocity_scale_input.value, color="black", label=vslab)
                # entry_layers.extend(scale_layers)

        # Store all layers for this entry in the registry
        layer_registry[entry_key] = entry_layers

site_circle_submit.on_click(site_circle)


def nn_graph(b):
    """Displays the 10 nearest neighbouring sites in a table and bar chart.

    Shows a table of site names, distances, velocities, velocity sigmas, and
    relative velocities in nearest_site_output, and a bar chart of distances
    in graph_output.

    Args:
        b: button click event, not used directly.
    """
    nearest_site_output.clear_output()
    graph_output.clear_output()

    # Parse map_radius_list into a dict keyed by site ID
    selected_options = {}
    for option in list(map_radius_list.value):
        option = option.split(", ")
        selected_options[option[0]] = {"org": option[1], "radius": int(option[2][0:-2])}

    for site in selected_options:
        if selected_options[site]["radius"] > 0:
            siteid = site
            rad = selected_options[site]["radius"]
            org = selected_options[site]["org"]
            site_list, _ = nearby_sites(org, siteid, rad)
            neighbours = sorted(site_list, key=lambda x: x[3])[:10]
            velocities = []; relvel = []; relsig = []
            refvel = data_of[org][siteid]["velocity"]
            refsig = data_of[org][siteid].get("velsig", [0, 0, 0])

            for site in neighbours:
                if isinstance(get_vel(site[0], org), tuple):
                    vel, sig = get_vel(site[0], org)
                    velocities.append((vel, sig))
                    relvel.append(vec_sub(vel, refvel))
                    # Difference of two estimates: the VARIANCES add. Assumes
                    # independence, so for neighbours these are upper bounds.
                    relsig.append([math.hypot(sig[i], refsig[i]) for i in range(3)])
                else:
                    velocities.append(("None", "None"))
                    relsig.append("None")
                    relvel.append("None")

            with nearest_site_output:
                display(wg.HTML("""<h2>10 Nearest Neighbouring Sites:</h2>"""))
                tableHTML = f"<table><tr><th><b>Site</b></th><th><b>Distance from {siteid}</b></th><th><b>Velocity (n, e, u)</b></th><th><b>VelSig (n, e, u)</b></th><th><b>RelVel (n, e, u)</b></th><th><b>RelSig (n, e, u)</b></th></tr>"
                for i in range(len(neighbours)):
                    if velocities[i][0] != "None":
                        velstr = [f"{x:.2f}" for x in velocities[i][0]]; velflt = [float(x) for x in velstr]
                        sigstr = [f"{x:.2f}" for x in velocities[i][1]]; sigflt = [float(x) for x in sigstr]
                        dvelstr = [f"{x:.2f}" for x in relvel[i]]; dvelflt = [float(x) for x in dvelstr]
                        dsigstr = [f"{x:.2f}" for x in relsig[i]]; dsigflt = [float(x) for x in dsigstr]
                    else:
                        velflt = "None"; sigflt = "None"; dvelflt = "None"; dsigflt = "None"
                    tableHTML += f"<tr><td>{neighbours[i][0]}</td><td>{neighbours[i][3]:.3f} km</td><td>{velflt}</td><td>{sigflt}</td><td>{dvelflt}</td><td>{dsigflt}</td></tr>"
                display(wg.HTML(tableHTML + "</table>"))

            s = str(plt.get_backend())
            if s.find('inline') > 0:
                with graph_output:
                    display(wg.HTML("""<h2 style="text-align:center;">Distances to Nearby Sites (km)</h2>"""))
                    plt.bar([x[0] for x in neighbours], [x[3] for x in neighbours])
                    plt.show()
            else:
                plt.figure(figsize=(8, 6))
                plt.bar([x[0] for x in neighbours], [x[3] for x in neighbours])
                plt.show()

plot_graph_butt.on_click(nn_graph)


def get_vel(siteid, org):
    """Retrieves velocity and velocity sigma for a site.

    Args:
        siteid: 4-character GNSS station identifier.
        org: organization key in data_of (e.g. "NGF", "UNR", "JPL").

    Returns:
        A tuple of (velocity, velsig) where velocity is [Vn, Ve, Vu] in
        mm/yr and velsig is [σVn, σVe, σVu] in mm/yr, or None if the
        site has no velocity data.
    """
    try:
        return data_of[org][siteid]["velocity"], data_of[org][siteid]["velsig"]
    except:
        return None


def plot_velocity(siteid, org, siglim):
    """Draws a velocity vector and optionally a confidence error ellipse.

    Only draws if all three velocity sigma components are within siglim.
    The ellipse is only drawn if plot_ellipse_check is also checked. Color
    encodes vertical velocity where blue is upward and red is downward, with
    saturation controlled by colorscalefactor_input. If plot_rel_vel is
    checked, the North and East components are plotted relative to the last
    reference site selected in map_radius_list, or to zero if nothing is
    selected; the vertical component is never differenced, and the colour
    always encodes the absolute up rate. In relative mode the ellipse uses
    the root sum of squares of the two stations' sigmas.

    Args:
        siteid: 4-character GNSS station identifier.
        org: organization key in data_of (e.g. "NGF", "UNR", "JPL").
        siglim: [σN_max, σE_max, σU_max] threshold in mm/yr. The vector and
            ellipse are only drawn if all three sigmas are within these limits.

    Returns:
        A list of ipyleaflet layers added to the map (vector arrow and
        optionally the error ellipse), for use by the layer registry.
    """
    layers = []
    if get_vel(siteid, org):
        velocity, velsig = get_vel(siteid, org)
    else:
        return layers

    refvel = [0, 0, 0]
    refsig = [0, 0, 0]
    refcorr = 0.0
    if plot_rel_vel.value:
        for option in list(map_radius_list.value):
            option = option.split(", ")
            site_ref = option[0]
            refvel = data_of[org][site_ref]["velocity"]
            refsig = data_of[org][site_ref].get("velsig", [0, 0, 0])
            refcorr = float(data_of[org][site_ref].get("velcorr", [0.0])[0])

    pltvel = [velocity[0]-refvel[0], velocity[1]-refvel[1], velocity[2]]
    # A difference of two estimates carries the SUM of their covariances, so
    # the ellipse for a relative vector combines both sigmas and both N/E
    # correlations. The two stations are assumed independent; neighbouring
    # stations share common-mode error, so these are upper bounds. The
    # vertical component is not differenced, so its sigma is unchanged.
    pltsig = [math.hypot(velsig[0], refsig[0]),
              math.hypot(velsig[1], refsig[1]), velsig[2]]
    sitecorr = float(data_of[org][siteid].get("velcorr", [0.0])[0])
    cov_ne = sitecorr*velsig[0]*velsig[1] + refcorr*refsig[0]*refsig[1]
    pltcorr = cov_ne/(pltsig[0]*pltsig[1]) if pltsig[0]*pltsig[1] else 0.0

    # Vertical velocity color encoding:
    # positive (upward) → blue: rgb(0, 0, bv)
    # negative (downward) → red: rgb(rv, 0, 0)
    # magnitude saturates at colorscalefactor_input mm/yr
    colorscalefactor = 255 / colorscalefactor_input.value

    if all(velsig[i] <= siglim[i] for i in range(3)):
        # Coloured from the ABSOLUTE up rate even in relative mode, because
        # pltvel[2] is deliberately not differenced (see the docstring).
        zv = velocity[2]
        if zv > 0:
            rv, bv = 0, min(255, zv * colorscalefactor)
        else:
            rv, bv = min(255, zv * (-colorscalefactor)), 0

        endpoint, vec_layers = draw_vector(
            map[0], data_of[org][siteid]["location"], pltvel,
            velocity_scale_input.value,
            color=f"rgb({int(rv)},0,{int(bv)})")
        layers.extend(vec_layers)

        # Draw error ellipse at arrow tip if ellipse checkbox is checked
        # Uses endpoint returned by draw_vector — not the station dot location
        if plot_ellipse_check.value:
            ellipse_layer = draw_ellipse(
                map[0], endpoint,
                pltsig[0], pltsig[1],
                ellipse_confidence.value,
                velocity_scale_input.value * ellipse_scale_input.value,
                color=f"rgb({int(rv)},0,{int(bv)})",
                correlation=pltcorr)
            if ellipse_layer is not None:
                layers.append(ellipse_layer)

    return layers


def draw_length(x):
    """Draws a scale bar line on the map of a given length in mm/yr.

    Currently inactive — draw_length_button.on_click is commented out.

    Args:
        x: button click event, not used directly.
    """
    length = arrow_length_input.value
    coords = length_coords_text.value.split(",")
    for i in range(len(coords)):
        coords[i] = float(coords[i])
    end_coords = calc_distance_earth(coords[0], coords[1], length)
    line = ipl.Polyline(locations=[coords, end_coords], weight=3)
    map[0].add_layer(line)

# draw_length_button.on_click(draw_length)


def clear_map_list(b):
    """Clears all entries from map_radius_list and removes all tracked layers.

    Uses the layer registry to remove all ipyleaflet layers from the map,
    then resets the site info panel.

    Args:
        b: button click event, not used directly.
    """
    for entry_key in list(layer_registry.keys()):
        for layer in layer_registry[entry_key]:
            try:
                map[0].remove_layer(layer)
            except Exception:
                pass
    layer_registry.clear()
    map_radius_list.options = []
    _site_info_html.value = ""
    # Reset named dict keys rather than reassigning the whole dict
    _current_site["site_id"] = None
    _current_site["org"] = None

clear_map_list_butt.on_click(clear_map_list)
```

## Using the Map

Clear/New Map, Reload Map, Close Map: Self Explanatory

Site ID + Source Dropdown + Radius + Add Sites: Inputting the combination of a valid site id, a radius, selecting a source, and then clicking "Add Sites" will add the format of "Siteid (Source) Radius km" to the Site/Radius Box. 

Site/Radius Box + Add Plot!: Here, select as many options as wished to be added to the map with the "Add Plot!" Button. This will show the site as well as all other sites within the selected radius. Other functions will also only work if the relevant site is selected from this box. 

Remove Sites: This will remove the Site/Source/Radius combination(s) from the Site/Radius Box

Clear All: Clears all sites from the Site/Radius box

Distance Table: This will show a table with the distances of the 10 nearest neighboring sites to the center site. 
It will also display their velocities, velocity relative to the first site, the velocity sigmas.  

Decimate: Thins out velocity vectors on the map. The number indicates how many 1 in x vectors will be shown; so if it's 10, then 1 in 10 vectors will be displayed on the map. 

Plot Velocities: Plots lines showing velocities; Relative Vel: Shows the North and East velocity relative to the selected site. The vertical component is never differenced, and the vector colour always shows the absolute up rate.

Arrow mm/yr: Set length velocity scale arrow.

Vel Scale: Sets the scale of velocities projection on the map.  Nominally, it relates mm/yr to kilometers on the map, i.e., 10 sets 1 mm/yr to be 10 km on the map.  Arrows, arrowheads and ellipses are all placed on the WGS-84 ellipsoid with the same geodesic used for the distance table.  The radius circle is drawn by Leaflet on its own sphere, so a station within about 125 m of a 50 km boundary may render on the opposite side of the circle from where the distance table puts it.  

<span style="color:red">The color of the vectors shows vertical motions with reds being down, blues up, and black near zero.</span>.  
±U Rate: Set the up velocity when the color saturates

Error Ellipses: Draws the confidence ellipse of the horizontal velocity at the arrow tip.  Ellipse Confidence sets the level; these are two-dimensional coverages, so 1σ, 2σ and 3σ enclose 39.35%, 86.47% and 98.89%, not the one-dimensional 68/95/99.7.  In Relative Vel mode the ellipse combines both stations' uncertainties, which assumes the two are independent; neighbouring stations share common-mode error, so a relative ellipse is an upper bound.  NGF publishes a North/East correlation, so NGF ellipses can tilt; UNR and JPL do not publish one, so theirs stay aligned to North and East.

Ellipse ×: Leave this at 1.  It is a deliberate exaggeration factor, and only at 1 are the ellipse and the arrow drawn at the same scale — which is the whole reason the ellipse can be read as the uncertainty of that arrow.  Note that two overlapping ellipses do <b>not</b> mean the two velocities are indistinguishable; testing a difference requires the uncertainty of the difference, which is larger than either ellipse.

Base Map Choice sets the base map with topography, street map, or ArcGIS imagery.

SigLim: The limits on the standard deviations of the points of the time-series plot expressed in (north, east, up). Sigma must be less than the values of each component. 

Detrended output: σVn, σVe and σVu are formal fit errors scaled by χ (NRMS). They assume white noise only; GNSS series carry flicker and random-walk noise this straight-line model does not represent, so these stay several times smaller than the velocity uncertainties the data centres publish and should not be quoted as the uncertainty of a published velocity. If breaks are removed, the offset uncertainties are not propagated: an offset error is shared by the whole post-break segment rather than being independent noise per epoch, and representing it properly needs a model parameter this fit does not have.

Remove N σ: Only works if Detrend is also checked. Removes outlier data from the graphs. The number determines how strict the outlier determination is; all data outside the standard deviation multiplied by the Remove N σ number will be removed. The data is then detrended and run through the process again until the beginning and ending number of data points remain the same. The number of removed data points and the number of iterations will be given.
